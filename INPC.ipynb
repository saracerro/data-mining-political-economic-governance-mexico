```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("INPC.csv",sep=';',dtype=str, encoding="utf-8")
df = df.replace(["N/E", " ", ""], np.nan)

for col in df.columns[1:]:
    df[col] = df[col].astype(str).str.replace(",", ".", regex=False)
    df[col] = df[col].astype(float)
df = df.rename(columns={df.columns[0]: "Fecha"})
print(df)
```

```python
def parse_fecha(s):
    if pd.isna(s):
        return pd.NaT
    s = s.strip()
    s = s.replace('-', ' ')  # per sicurezza
    parts = s.split()
    if len(parts) < 3:
        return pd.NaT
    quincena, mes, año = parts[:3]
    meses = {
        "Ene": 1, "Feb": 2, "Mar": 3, "Abr": 4, "May": 5, "Jun": 6,
        "Jul": 7, "Ago": 8, "Sep": 9, "Oct": 10, "Nov": 11, "Dic": 12
    }
    mes_num = meses.get(mes, 1)
    dia = 1 if "1Q" in quincena else 15
    return pd.Timestamp(year=int(año), month=mes_num, day=dia)

```

```python

df["Fecha"] = df["Fecha"].apply(parse_fecha)
df = df.set_index("Fecha").sort_index()
```

```python
df_reset = df.reset_index()
df_long = df_reset.melt(
    id_vars="Fecha",               # la colonna da mantenere (ex-index)
    var_name="codigo_ciudad",      # nome della colonna per i codici città
    value_name="INPC"         # nome della colonna dei valori
)

df_long["codigo_ciudad"] = df_long["codigo_ciudad"].astype(int)

df_long["STATO"] = df_long["codigo_ciudad"].map(
    lambda x: codigo_a_ciudad[x]["estado"] if x in codigo_a_ciudad else np.nan
)
print(df_long["INPC"].isna().sum())
df_long = df_long.sort_values(["Fecha", "codigo_ciudad"]).reset_index(drop=True)
```

```python
def inflacion_anual(grupo):
    # Crea una colonna "ANNO"
    grupo["ANNO"] = grupo["Fecha"].dt.year

    risultati = []

    for anno, dati_anno in grupo.groupby("ANNO"):
        if anno == 2000:
            # Per il 2000 → inflazione = 0
            risultati.append({"ANNO": anno, "INFLACION": 0})
            continue

        # Trova il valore più vicino a 1 gennaio e 15 dicembre di quell’anno
        inizio = dati_anno.iloc[(dati_anno["Fecha"] - pd.Timestamp(f"{anno}-01-01")).abs().argsort()[:1]]
        fine = dati_anno.iloc[(dati_anno["Fecha"] - pd.Timestamp(f"{anno}-12-15")).abs().argsort()[:1]]

        # Ottieni i valori numerici (o NaN)
        inpc_inizio = inizio["INPC"].values[0] if not inizio.empty else np.nan
        inpc_fine = fine["INPC"].values[0] if not fine.empty else np.nan

        # Se uno dei due è NaN → risultato NaN
        if pd.isna(inpc_inizio) or pd.isna(inpc_fine):
            inflacion = np.nan
        else:
            inflacion = (inpc_fine - inpc_inizio) / inpc_inizio * 100

        risultati.append({"ANNO": anno, "INFLACION": inflacion})

    return pd.DataFrame(risultati)
```

```python
df_long["Fecha"] = pd.to_datetime(df_long["Fecha"])
inflacion_anual_ciudad = (
    df_long.groupby("codigo_ciudad", group_keys=True)
    .apply(inflacion_anual)
    .reset_index()
)
inflacion_anual_ciudad = inflacion_anual_ciudad.drop(columns=["level_1"])


```

```python
def media_stato(grupo):
    if grupo["INFLACION"].isna().any():
        return np.nan
    else:
        return grupo["INFLACION"].mean()
```

```python
inflacion_anual_estado = (
    inflacion_anual_ciudad.groupby(["ANNO", "STATO"])
    .apply(media_stato)
    .reset_index(name="INFLACION_ANUAL_ESTADO")
)

print(inflacion_anual_estado.dtypes)
inflacion_anual_estado.to_csv("inflacion_anual_estado.csv", index=False)
```
