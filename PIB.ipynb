```python
import pandas as pd
import numpy as np
pib1=pd.read_csv('PIBE_r_1.csv', sep=';', encoding='utf8', low_memory=False, skiprows=[0,1,2,3,5,6], nrows=34, usecols=['Concepto', '2000','2001','2002','2003','2004','2005', '2006','2007','2008', '2009','2010','2011', '2012','2013','2014', '2015','2016','2017', '2018','2019','2020','2021'] )
pib2=pd.read_csv('PIBE_r_3.csv', sep=';', encoding='utf8', low_memory=False, skiprows = [0, 1, 2, 3] + list(range(5, 146)),nrows=34,  usecols=['Concepto', '2000','2001','2002','2003','2004','2005', '2006','2007','2008', '2009','2010','2011', '2012','2013','2014', '2015','2016','2017', '2018','2019','2020','2021'] )
# Assicurati che le colonne 2022-2024 esistano
for col in ['2022','2023','2024']:
    if col not in pib1.columns:
        pib1[col] = np.nan
    if col not in pib2.columns:
        pib2[col] = np.nan

pib1['Concepto'] = pib1['Concepto'].str.upper()

pib2['Concepto'] = pib2['Concepto'].str.upper()


# --- Lista anni ---
anni = [str(i) for i in range(2000, 2025)]

# --- Trasforma pib1 in formato lungo ---
pib1_long = pib1.melt(
    id_vars="Concepto",
    value_vars=anni,
    var_name="ANNO",
    value_name="PIB"
)
pib1_long["ANNO"] = pib1_long["ANNO"].astype(int)
pib1_long = pib1_long.rename(columns={"Concepto": "ENTIDAD"})

# --- Trasforma pib2 in formato lungo ---
pib2_long = pib2.melt(
    id_vars="Concepto",
    value_vars=anni,
    var_name="ANNO",
    value_name="PIB_GROWTH"
)

print(pib1_long)
print(pib2_long)

pib2_long["ANNO"] = pib2_long["ANNO"].astype(int)
pib2_long = pib2_long.rename(columns={"Concepto": "ENTIDAD"})
pib1_long = pd.merge(
    pib1_long,
    pib2_long[['ENTIDAD','ANNO','PIB_GROWTH']],
    how='left',
    on=['ENTIDAD','ANNO']
)
print(pib1_long)
pib1_long.to_csv("pib_estatal.csv", index=False, encoding="latin_1", sep=',')
```
